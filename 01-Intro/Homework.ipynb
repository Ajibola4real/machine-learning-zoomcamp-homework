```python
df_filtered = df_filtered.head(7)
```

```python
X = df_filtered.to_numpy()
X
```

```python
XTX = X.T @ (X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

y 

```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```
